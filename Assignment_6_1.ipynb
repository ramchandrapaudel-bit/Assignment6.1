# Assignment 6.1 Decision Trees, Random Forests

 ## Activity 4.01: Car Data Classification

```python
# Step 1: Import libraries
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
```

```python
# Step 2: Load the dataset
url = (
    "https://archive.ics.uci.edu/ml/"
    "machine-learning-databases/car/car.data"
)

column_names = [
    "buying",
    "maintenance",
    "doors",
    "persons",
    "luggage_boot",
    "safety",
    "class"
]

df = pd.read_csv(url, names=column_names)

print("First five rows of the dataset:")
print(df.head())

print("\nDataset shape:")
print(df.shape)
```

```python
# Step 3: Encode the categorical columns
feature_columns = [
    "buying",
    "maintenance",
    "doors",
    "persons",
    "luggage_boot",
    "safety"
]

feature_encoders = {}

for column in feature_columns:
    feature_encoders[column] = LabelEncoder()
    df[column] = feature_encoders[column].fit_transform(df[column])

# Encode the target separately
target_encoder = LabelEncoder()
df["class"] = target_encoder.fit_transform(df["class"])
```

```python
# Step 4: Extract the target using pop()
labels = df.pop("class")
features = df

print("\nEncoded features:")
print(features.head())

print("\nEncoded target labels:")
print(labels.head())
```

```python
# Step 5: Split into training and testing data
features_train, features_test, labels_train, labels_test = (
    train_test_split(
        features,
        labels,
        test_size=0.10,
        random_state=42,
        stratify=labels
    )
)

print("\nTraining samples:", len(features_train))
print("Testing samples:", len(features_test))
```

```python
# Step 6: Create and train the decision tree
tree_classifier = DecisionTreeClassifier(random_state=42)

tree_classifier.fit(features_train, labels_train)
```

```python
# Step 7: Predict the test labels
labels_predicted = tree_classifier.predict(features_test)

# Evaluate using score()
accuracy = tree_classifier.score(features_test, labels_test)

print("\nDecision Tree Test Accuracy:")
print(f"{accuracy:.4f} ({accuracy * 100:.2f}%)")
```

```python
# Step 8: Print the classification report
print("\nClassification Report:")

print(
    classification_report(
        labels_test,
        labels_predicted,
        labels=range(len(target_encoder.classes_)),
        target_names=target_encoder.classes_,
        zero_division=0
    )
)
```

## Activity 4.02: Random Forest Classification for Your Car Rental Company

```python
# Step 1: Import libraries
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
# Step 2: Load the Car Evaluation dataset
url = (
    "https://archive.ics.uci.edu/ml/"
    "machine-learning-databases/car/car.data"
)

column_names = [
    "buying",
    "maintenance",
    "doors",
    "persons",
    "luggage_boot",
    "safety",
    "class"
]

df = pd.read_csv(url, names=column_names)

print("Dataset shape:", df.shape)
```

```python
# Step 3: Encode the categorical features
feature_columns = [
    "buying",
    "maintenance",
    "doors",
    "persons",
    "luggage_boot",
    "safety"
]

feature_encoders = {}

for column in feature_columns:
    feature_encoders[column] = LabelEncoder()
    df[column] = feature_encoders[column].fit_transform(df[column])

# Encode the target separately
target_encoder = LabelEncoder()
df["class"] = target_encoder.fit_transform(df["class"])
```

```python
# Step 4: Extract target using pop()
labels = df.pop("class")
features = df
```

```python


# Step 5: Split the dataset (90% training, 10% testing)
features_train, features_test, labels_train, labels_test = (
    train_test_split(
        features,
        labels,
        test_size=0.10,
        random_state=42,
        stratify=labels
    )
)

# Function to train and evaluate each classifier
def evaluate_model(model, model_name):

    # Train the model
    model.fit(features_train, labels_train)

    # Make predictions
    predictions = model.predict(features_test)

    # Display model accuracy
    print("\n" + "=" * 55)
    print(model_name)
    print("=" * 55)

    print(
        f"Test Accuracy: "
        f"{accuracy_score(labels_test, predictions):.4f}"
    )

    # Display confusion matrix
    print("\nConfusion Matrix:")

    matrix = confusion_matrix(
        labels_test,
        predictions,
        labels=range(len(target_encoder.classes_))
    )

    print(matrix)

    # Display classification report
    print("\nClassification Report:")

    print(
        classification_report(
            labels_test,
            predictions,
            labels=range(len(target_encoder.classes_)),
            target_names=target_encoder.classes_,
            zero_division=0
        )
    )

    # Display feature importance
    print("\nFeature Importance:")

    importance = pd.Series(
        model.feature_importances_,
        index=features.columns,
        name="Importance"
    ).sort_values(ascending=False)

    print(importance)

    return model






```

```python
# Step 6: Create and evaluate Random Forest
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

random_forest = evaluate_model(
    random_forest,
    "Random Forest Classifier"
)
```

```python
# Step 7: Create and evaluate Extra Trees
extra_trees = ExtraTreesClassifier(
    n_estimators=100,
    random_state=42
)

extra_trees = evaluate_model(
    extra_trees,
    "Extra Trees Classifier"
)
```

```python
# Final comparison of the two models
print("\n" + "=" * 55)
print("FINAL MODEL COMPARISON")
print("=" * 55)

rf_accuracy = random_forest.score(features_test, labels_test)
et_accuracy = extra_trees.score(features_test, labels_test)

print(f"Random Forest Accuracy: {rf_accuracy:.4f}")
print(f"Extra Trees Accuracy:   {et_accuracy:.4f}")

if rf_accuracy > et_accuracy:
    print("Higher test accuracy: Random Forest")
elif et_accuracy > rf_accuracy:
    print("Higher test accuracy: Extra Trees")
else:
    print("Both models have equal test accuracy.")
```
